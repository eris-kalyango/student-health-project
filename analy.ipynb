```python
sample1 = df["Age"].sample(n=100,random_state=42)
sample1.mean()
```

```python
from scipy.stats import ttest_1samp
t_stat,p_value = ttest_1samp(sample1,popmean=50)

print("t-statistic:",t_stat)
print("p-value:",p_value)
```

```python
import numpy as np
mean = df["Age"].mean()
std = df["Age"].std()

n = len(df)
lower_bound = mean - 1.96 * (std / np.sqrt(n))
upper_bound = mean + 1.96 * (std / np.sqrt(n))

print(f"THE CI IS {lower_bound} to {upper_bound}")
```
